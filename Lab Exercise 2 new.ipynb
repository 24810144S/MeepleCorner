## Question 2

The students got the following marks towards the semester's end.

![namelist](https://github.com/ychen9295/FinTech-and-Blockchain/blob/main/images/lab_exercise03_marks.png?raw=1)

- Write code to update the marks in `data` according to the above table.
- Add a column named 'total' in the DataFrame `results` with the overall marks for individual students
- Write code to find out which student got the highest mark in this course.
- Write code to find out which student got the lowest mark in this course.



```python
# Question 2

# Write your code below:

# ─── 1. Update the marks according to the table ──────────────────
results["HW1"] = [81, 71, 99, 52, 42, 76, 71, 31, 69, 39, 68, 34, 43, 95, 81, 34, 100, 65, 74, 31]
results["HW2"] = [88, 49, 36, 65, 34, 61, 44, 100, 86, 85, 57, 93, 65, 87, 99, 71, 73, 96, 63, 85]
results["Test"] = [98, 54, 32, 63, 98, 54, 56, 33, 51, 90, 61, 89, 48, 90, 39, 79, 74, 49, 73, 52]
results["Exam"] = [38, 54, 100, 86, 92, 64, 88, 72, 82, 55, 74, 93, 64, 81, 41, 36, 98, 56, 79, 37]

# ─── 2. Add a 'total' column with weighted overall marks ─────────
# Total = (HW1 * 0.2) + (HW2 * 0.2) + (Test * 0.2) + (Exam * 0.4)
results["total"] = (results["HW1"] * 0.2) + (results["HW2"] * 0.2) + (results["Test"] * 0.2) + (results["Exam"] * 0.4)

# ─── 3. Find the student with the highest mark ───────────────────
# .idxmax() returns the index (student name) of the highest value
highest_student = results["total"].idxmax()
highest_mark = results["total"].max()

# ─── 4. Find the student with the lowest mark ────────────────────
# .idxmin() returns the index (student name) of the lowest value
lowest_student = results["total"].idxmin()
lowest_mark = results["total"].min()

# ─── 5. Display the Results ──────────────────────────────────────
print("--- Updated DataFrame ---")
print(results)
print("\n--- Analysis ---")
print(f"Highest mark: {highest_student} ({highest_mark})")
print(f"Lowest mark: {lowest_student} ({lowest_mark})")
```
