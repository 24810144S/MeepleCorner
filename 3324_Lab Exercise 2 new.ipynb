```python
# Question 2

# Write your code below:

# Update the marks according to the table
results["HW1"] = [81, 71, 99, 52, 42, 76, 71, 31, 69, 39, 68, 34, 43, 95, 81, 34, 100, 65, 74, 31]
results["HW2"] = [88, 49, 36, 65, 34, 61, 44, 100, 86, 85, 57, 93, 65, 87, 99, 71, 73, 96, 63, 85]
results["Test"] = [98, 54, 32, 63, 98, 54, 56, 33, 51, 90, 61, 89, 48, 90, 39, 79, 74, 49, 73, 52]
results["Exam"] = [38, 54, 100, 86, 92, 64, 88, 72, 82, 55, 74, 93, 64, 81, 41, 36, 98, 56, 79, 37]

# Add a 'total' column with weighted overall marks

results["total"] = (results["HW1"] * 0.2) + (results["HW2"] * 0.2) + (results["Test"] * 0.2) + (results["Exam"] * 0.4)

# Find the student with the highest mark
highest_student = results["total"].idxmax()
highest_mark = results["total"].max()

# Find the student with the lowest mark
lowest_student = results["total"].idxmin()
lowest_mark = results["total"].min()

print("--- Updated DataFrame ---")
print(results)
print("\n--- Analysis ---")
print(f"Highest mark: {highest_student} ({highest_mark})")
print(f"Lowest mark: {lowest_student} ({lowest_mark})")
```

## Question 3

Suppose the following mark-to-grade conversion table is adopted in this course.

|Mark range| Grade|
|:-: | :-: |
|49 or below | F |
|50 to 59| D|
|60 to 69| C|
|70 to 79| B|
|80 or above| A|

- Add a column named 'grade' in the DataFrame `results` with the overall grade for individual students.




```python
# Question 3

# Write your code below:

# Define a function that maps the mark to the grade based on the table
def assign_grade(mark):
    if mark >= 80:
        return 'A'
    elif mark >= 70:
        return 'B'
    elif mark >= 60:
        return 'C'
    elif mark >= 50:
        return 'D'
    else:
        return 'F'

# Apply the function to the 'total' column to create the new 'grade' column
results['grade'] = results['total'].apply(assign_grade)

results
```
